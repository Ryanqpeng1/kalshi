# Kalshi Trading Dashboard

Interactive notebook for managing trades on Kalshi prediction markets.
- View available markets and their spreads
- Place buy/sell orders
- Monitor current positions and account balance

```python
# Import Required Libraries
import pandas as pd
import numpy as np
from datetime import datetime
import sys
import os

# Add current directory to path
sys.path.insert(0, os.getcwd())

# Import Kalshi modules
from lol import KalshiClient, KalshiConfig
from markets import get_market_positions, get_lol_positions
from config import API_KEY_ID, PRIVATE_KEY_PATH, ENVIRONMENT

pd.set_option('display.max_rows', None)

```

## 1. Initialize API Connection

Connect to Kalshi trading API using credentials from config

```python
# Initialize Kalshi API Client
config = KalshiConfig()
config.ENVIRONMENT = ENVIRONMENT
config.API_KEY_ID = API_KEY_ID
config.PRIVATE_KEY_PATH = PRIVATE_KEY_PATH

client = KalshiClient(config)

# Test connection and get balance
try:
    balance_result = client.get_balance()
    balance = balance_result.get('balance_in_cents', 0) / 100
    print(f"Connected to {ENVIRONMENT.upper()} environment")
    print(f"Account Balance: ${balance:.2f}")
except Exception as e:
    print(f"Connection error: {e}")
```

## 2. Display Market Information

Fetch and display all available markets with volume and bid-ask spreads

```python
markets_df = get_market_positions(volume_threshold=0.0, limit=1000)

display(markets_df)
```

## 3. Place an Order

Place a buy or sell order on a selected market

```python
# PLACE THE ORDER
print("Placing order...")
print(f"  Ticker: {TICKER}")
print(f"  Side: {SIDE.upper()}")
print(f"  Action: {ACTION.upper()}")
print(f"  Count: {COUNT}")
print(f"  Price: ${PRICE/100:.2f}")

try:
    order_result = client.place_order(
        ticker=TICKER,
        side=SIDE,
        action=ACTION,
        count=COUNT,
        price=PRICE
    )
    
    if isinstance(order_result, dict):
        order_id = order_result.get('order_id', order_result)
    else:
        order_id = order_result
    
    print(f"\nOrder placed successfully!")
    print(f"  Order ID: {order_id}")
    
except Exception as e:
    print(f"\nError placing order: {e}")
    print("  Check that ticker is valid and you have sufficient balance")
```

## 4. Check Current Positions

View all open positions and account balance

```python
# Get account balance
try:
    balance_result = client.get_balance()
    balance_cents = balance_result.get('balance_in_cents', 0)
    balance_dollars = balance_cents / 100
    
    print("="*100)
    print(f"Account Balance: ${balance_dollars:,.2f}")
    print("="*100)
except Exception as e:
    print(f"Error fetching balance: {e}")

# Get current positions
try:
    positions = client.get_positions()
    
    if positions:
        print(f"\nCurrent Positions: {len(positions)} holdings\n")
        
        # Convert to DataFrame
        positions_df = pd.DataFrame(positions)
        
        # Remove columns/rows with all NaN values
        positions_df = positions_df.dropna(axis=1, how='all')
        
        # Display relevant columns (adapt to available fields)
        display_columns = ['ticker', 'side', 'quantity', 'average_entry_price', 'count', 'fee_paid']
        available_cols = [c for c in display_columns if c in positions_df.columns]
        
        if available_cols:
            print(positions_df[available_cols].to_string())
        else:
            print("Available columns:", list(positions_df.columns))
            print(positions_df.to_string())
        
        # Summary
        if 'quantity' in positions_df.columns:
            total_contracts = positions_df['quantity'].sum()
        elif 'count' in positions_df.columns:
            total_contracts = positions_df['count'].sum()
        else:
            total_contracts = len(positions_df)
            
        print(f"\nTotal contracts held: {total_contracts}")
    else:
        print("No open positions")
        
except Exception as e:
    print(f"Error fetching positions: {e}")
    import traceback
    traceback.print_exc()
```

## 5. Portfolio Summary & Analysis

Quick statistics on your trading activity

```python
# Portfolio Analysis
print("="*100)
print("Portfolio Summary")
print("="*100)

print(f"\nAccount Balance: ${balance_dollars:,.2f}")

# Top spread opportunities
if not markets_df.empty:
    widest = markets_df.copy()
    widest['avg_spread'] = (widest['yes_spread'] + widest['no_spread']) / 2
    
    print(f"\nMarkets with Best Spreads (Top 3):")
    for idx, row in widest.nlargest(3, 'avg_spread').iterrows():
        print(f"  • {row['title'][:40]}")
        print(f"    YES Spread: ${row['yes_spread']:.2f} | NO Spread: ${row['no_spread']:.2f}")

# Trading statistics
print(f"\nMarket Statistics:")
print(f"  Total markets: {len(markets_df)}")
print(f"  Markets with volume > 0: {len(markets_df[markets_df['volume'] > 0])}")
print(f"  Average volume: {markets_df['volume'].mean():.1f}")
print(f"  Max volume: {markets_df['volume'].max():.1f}")

# Latest update timestamp
print(f"\nData updated at: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
```
